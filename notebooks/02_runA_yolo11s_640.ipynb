```python
from ultralytics import YOLO

model = YOLO("yolo11s.pt")
model.train(
    data="VisDrone.yaml",
    epochs=30,
    imgsz=640,
    batch=16,
    device=0,
    seed=0,
    deterministic=False,
    workers=4,
    project="/kaggle/working/runs",
    name="yolo11s_640",
    exist_ok=True,
)
print("RUN A FINISHED")
```
